```python
import numpy as np

M = np.array([
    [4, -2, 7, 0],
    [1, 9, 3, -5],
    [8, 6, -1, 2]
])

m, n = M.shape
print(f"Matrix Dimensions: {m} x {n}")

print("m_1,3 =", M[0, 2])
print("m_2,4 =", M[1, 3])
print("m_3,2 =", M[2, 1])
```

```python
import numpy as np


A = np.array([[3, 5], [-1, 4]])
B = np.array([[0, -2], [7, 1]])
C = np.array([[1, 2, 3], [4, 5, 6]])


A_plus_B = A + B
print("1. A + B =\n", A_plus_B)


A_minus_B = A - B
print("\n2. A - B =\n", A_minus_B)


try:
    A_plus_C = A + C
except ValueError as e:
    print("\n3. Is A + C possible? No!")
    print("Reason:", e)
```

```python
import numpy as np


X = np.array([[2, -1], [3, 0]])
Y = np.array([[1, 4], [-2, 5]])


scalar_mult = 3 * X
print("1. 3X =\n", scalar_mult)

element_wise = X * Y
print("\n2. Element-wise (X * Y) =\n", element_wise)


dot_product = X @ Y  # أو np.dot(X, Y)
print("\n3. Dot Product (X · Y) =\n", dot_product)
```

```python
import numpy as np

K = np.array([[5, 2], [3, 4]])

K_transpose = K.T
print("1. Transpose (K^T) =\n", K_transpose)

K_det = np.linalg.det(K)
print(f"\n2. Determinant det(K) = {round(K_det)}")

K_inverse = np.linalg.inv(K)
print("\n3. Inverse (K^-1) =\n", K_inverse)
```

```python
import numpy as np

A = np.array([[2, 3], [4, -1]])
B = np.array([13, 5])

X = np.linalg.solve(A, B)

print("x =", round(X[0]))
print("y =", round(X[1]))
```

```python
import numpy as np
from scipy import stats

data = [120, 150, 130, 180, 150, 200, 110, 160]

mean_val = np.mean(data)
print("1. Mean (x_bar) =", mean_val)

median_val = np.median(data)
print("2. Median =", median_val)

mode_val = stats.mode(data, keepdims=True).mode[0]
print("3. Mode =", mode_val)
```

```python
import numpy as np

data = [2, 4, 6, 8, 10]

s = np.std(data, ddof=1)
print("1. Sample Standard Deviation (s) =", round(s, 2))

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

print("2. Q1 =", q1)
print("   Q3 =", q3)
print("   IQR =", iqr)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Area': [1200, np.nan, 1800],
    'Bedrooms': [2, 3, 4],
    'Price': [200000, 250000, 350000]
})

mean_area = df['Area'].mean()
df['Area'] = df['Area'].fillna(mean_area)
print("1. Imputed Area:\n", df['Area'])

area_min = df['Area'].min()
area_max = df['Area'].max()
df['Area_Scaled'] = (df['Area'] - area_min) / (area_max - area_min)

print("\n2. Scaled Area:\n", df['Area_Scaled'])
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = np.arange(100).reshape(-1, 1)

X_train, X_test = train_test_split(X, test_size=0.20, random_state=42)

print("X_train samples:", len(X_train))
print("X_test samples:", len(X_test))

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```
